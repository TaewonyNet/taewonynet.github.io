# 구글 코랩으로 드라이브에 파일을 내려받아보자

Google Drive를 마운트하고, requests 청크 스트림으로 파일을 Drive에 직접 저장합니다.
셀을 위에서 아래로 순서대로 실행하세요.

```python
# 1. Google Drive 마운트
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 2. 저장 폴더 설정
from pathlib import Path

SAVE_DIR = Path('/content/drive/MyDrive/downloads')
SAVE_DIR.mkdir(parents=True, exist_ok=True)
print(f'저장 폴더: {SAVE_DIR}')
```

```python
# 3. URL에서 Drive로 파일 내려받기
import requests

SAMPLE_URL = 'https://www.w3schools.com/html/mov_bbb.mp4'  # 받을 파일 URL 교체
FILENAME = 'sample.mp4'  # 저장 파일명 교체

dest = SAVE_DIR / FILENAME
with requests.get(SAMPLE_URL, stream=True, timeout=60) as r:
    r.raise_for_status()
    total = int(r.headers.get('content-length', 0))
    downloaded = 0
    with open(dest, 'wb') as f:
        for chunk in r.iter_content(chunk_size=1 * 1024 * 1024):
            f.write(chunk)
            downloaded += len(chunk)
            if total:
                print(f'\r{downloaded / total * 100:.1f}%  ({downloaded / 1024 / 1024:.1f} MB)', end='', flush=True)

print(f'\n저장 완료: {dest}  ({dest.stat().st_size / 1024 / 1024:.2f} MB)')
```

```python
# 4. Drive 저장 파일 목록 확인
print(f'{SAVE_DIR} 파일 목록:')
for p in sorted(SAVE_DIR.iterdir()):
    size_mb = p.stat().st_size / 1024 / 1024
    print(f'  {p.name:<40} {size_mb:.2f} MB')
```

```python
# 3. 저장 폴더 설정
from pathlib import Path

SAVE_DIR = Path('/content/drive/MyDrive/videos')
SAVE_DIR.mkdir(parents=True, exist_ok=True)
print(f'저장 폴더: {SAVE_DIR}')
```

```python
# 4. 동영상 내려받기
import yt_dlp

VIDEO_URL = 'https://www.youtube.com/watch?v=BaW_jenozKc'  # 받을 영상 URL 교체

opts = {
    'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]/best',
    'outtmpl': str(SAVE_DIR / '%(title)s.%(ext)s'),
    'merge_output_format': 'mp4',
}

with yt_dlp.YoutubeDL(opts) as ydl:
    info = ydl.extract_info(VIDEO_URL, download=True)
    title = info.get('title', 'unknown')
    duration = info.get('duration', 0)

print(f'\n완료: {title} ({duration}초)')
```

```python
# 5. Drive 저장 파일 목록 확인
print(f'{SAVE_DIR} 파일 목록:')
for p in sorted(SAVE_DIR.iterdir()):
    size_mb = p.stat().st_size / 1024 / 1024
    print(f'  {p.name:<60} {size_mb:.1f} MB')
```
